```python
# 5.3

n = int(input("Enter num"))
sum = 0

for i in range(2,n+1):
    isPrime = True

    for j in range(2,int(i**0.5)):
        if(i % j == 0):
            isPrime = False
            break;

    if isPrime == True:
        sum = sum + i

print(sum)
```
