#### Q.2 Questions on Basic NumPy Array:
##### a) Reverse the NumPy array: arr = np.array([1, 2, 3, 6, 4, 5])

```python
arr = np.array([1, 2, 3, 6, 4, 5])
print("Original:", arr)
print("Reversed:", arr[::-1])
```

##### b) Find the most frequent value and their indice(s) in the following arrays:
##### i. x = np.array([1,2,3,4,5,1,2,1,1,1])
##### ii. y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3 ])

```python
def most_frequent(arr):
    values, counts = np.unique(arr, return_counts=True)
    max_count = counts.max()
    for v in values[counts == max_count]:
        print(f"  Most frequent value: {v} (appears {max_count} times) at indices {np.where(arr == v)[0]}")

x = np.array([1,2,3,4,5,1,2,1,1,1])
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3 ])

print("i. x =", x)
most_frequent(x)
print("ii. y =", y)
most_frequent(y)
```

#### Q.4 Write program to create an 1-D NumPy array named <<Your Name>> with evenly spaced 25 numbers from 10 to 100 using linspace(). Print the dimensions of the array, shape, total elements, the data type of each element and total number of bytes consumed by the array. Find the transpose of this array using reshape() attribute. Can we do the same with T attribute?

```python
Paramveer = np.linspace(10, 100, 25)
print(Paramveer)
print("Dimensions (ndim)   :", Paramveer.ndim)
print("Shape               :", Paramveer.shape)
print("Total elements(size):", Paramveer.size)
print("Data type (dtype)   :", Paramveer.dtype)
print("Total bytes (nbytes):", Paramveer.nbytes)
```

```python
# Transpose using reshape(): turn the row (25,) into a column (25, 1)
transposed = Paramveer.reshape(25, 1)
print("Shape after reshape:", transposed.shape)
print(transposed)
```

```python
# Using the T attribute
print("Shape after .T:", Paramveer.T.shape)
print("Same as original?", np.array_equal(Paramveer.T, Paramveer))
```

**Answer:** No. A 1-D array has only one axis, so `.T` has nothing to swap and returns the array unchanged (shape stays `(25,)`). `.T` works as a transpose only for arrays with 2 or more dimensions. To get the column form with `.T`, first make the array 2-D, e.g. `yourname.reshape(1, 25).T` gives shape `(25, 1)`.

```python
print("2-D then .T:", Paramveer.reshape(1, 25).T.shape)
```

#### Q5. Create a 2-D Array of three rows and four columns, named ucs420_<your_name> with following values – 10, 20, 30, 40, 50, 60, 70, 80, 90, 15, 20, 35. Compute the mean, median, max, min, unique elements. Reshape the array to four rows and three columns and name it as reshaped_ ucs420_<your_name>. Resize the array to two rows and three columns and name it as resized_ ucs420_<your_name>.

```python
ucs420_Paramveer = np.array([10, 20, 30, 40, 50, 60, 70, 80, 90, 15, 20, 35]).reshape(3, 4)
print(ucs420_Paramveer)

print("Mean   :", np.mean(ucs420_Paramveer))
print("Median :", np.median(ucs420_Paramveer))
print("Max    :", np.max(ucs420_Paramveer))
print("Min    :", np.min(ucs420_Paramveer))
print("Unique :", np.unique(ucs420_Paramveer))
```

```python
reshaped_ucs420_Paramveer = ucs420_Paramveer.reshape(4, 3)
print("Reshaped (4 x 3):")
print(reshaped_ucs420_Paramveer)
```

```python
resized_ucs420_Paramveer = np.resize(ucs420_Paramveer, (2, 3))
print("Resized (2 x 3):")
print(resized_ucs420_Paramveer)
```
