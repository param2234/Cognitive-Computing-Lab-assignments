### 3. Set the random seed to your own roll number using random.seed(your_roll_number) before generating your numbers — this keeps your list random-looking but reproducible and unique to you (a different seed produces a different list, so this cannot be copied from a classmate).
##### i. Generate a list of 100 random numbers between 100 and 900 (inclusive).

```python
import random

roll_number = 1024170173
random.seed(roll_number)

nums = [random.randint(100, 900) for _ in range(100)]

print("Random Numbers:", nums)
```

##### ii. Count and print all odd numbers in the list.

```python
odd = [n for n in nums if n % 2 != 0]

print("Odd count:", len(odd))
```

##### iii. Count and print all even numbers in the list.

```python
even = [n for n in nums if n % 2 == 0]

print("Even count:", len(even))
```

##### iv. Count and print all prime numbers in the list, and additionally build the actual list of prime numbers found using list comprehension.

```python
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

prime_count = sum(1 for n in nums if is_prime(n))
primes = [n for n in nums if is_prime(n)]

print("Prime count:", prime_count)
print("Primes list:", primes)
```

##### v. Print the number that occurs most frequently in your list, and how many times it occurs.

```python
from collections import Counter

c = Counter(nums)
most_common_num, freq = c.most_common(1)[0]

print("Most frequent number:", most_common_num)
print("Frequency:", freq)
```

### 4. Using the same 8 digits of your roll number from Q1 (before multiplying by 10), create set A = {digit × 7 for each digit} and set B = {digit × 9 for each digit}. Print both sets.

```python
roll_number = "1024170173"
digits8 = roll_number[:8]

A = {int(d) * 7 for d in digits8}
B = {int(d) * 9 for d in digits8}

print("Set A:", A)
print("Set B:", B)
```

##### vi. Find the union of A and B — all unique values across both sets.

```python
print("Union:", A | B)
```

##### vii. Find the intersection of A and B — values common to both.

```python
print("Intersection:", A & B)
```

##### viii. Find A − B and B − A separately (not combined) using difference(), and explain in one line how this differs from symmetric_difference().

```python
print("A - B:", A.difference(B))
print("B - A:", B.difference(A))

# difference() only returns elements unique to the first set (one direction at a time),
# while symmetric_difference() combines both directions into a single set of all elements not shared by both.
```

##### ix. Find the symmetric difference of A and B.

```python
print("Symmetric Difference:", A.symmetric_difference(B))
```

##### x. Check whether A is a subset of B, and whether B is a superset of A, using issubset() and issuperset().

```python
print("Is A subset of B?", A.issubset(B))
print("Is B superset of A?", B.issuperset(A))
```

##### xi. Ask the user to input a value X. Remove X from set A if present using discard() (not remove()) — explain in one line why discard() is safer than remove() when you're not sure the value exists.

```python
X = int(input("Enter a value to discard from A: "))

A.discard(X)
print("Set A after discard:", A)

# discard() is safer than remove() because it does nothing if X is not in the set,
# whereas remove() raises a KeyError when the value is missing.
```

### 5. Consider the following dictionary:



```python
my_dict = {
    "name": "Aarav Sharma",
    "roll_no": "1024170173",
    "branch": "CSE",
    "age": 20,
    "city": "Patiala"
}

print(my_dict)
```

##### i. Rename the key “city” to “location” without changing its value (do not just recreate the dictionary by hand — use pop() or dictionary comprehension so the code would work for any dictionary).

##### iv. Delete the “branch” key using two different methods in two separate copies of the dictionary: once with pop() and once with the del keyword. Explain in one line how pop() differs from del (hint: what does each return?).

```python
dict_pop = my_dict.copy()
removed_value = dict_pop.pop("branch")
print("Removed with pop():", removed_value)
print("Dict after pop():", dict_pop)

dict_del = my_dict.copy()
del dict_del["branch"]
print("Dict after del:", dict_del)

# pop() removes the key and returns its value, so it can be stored in a variable,
# while del simply deletes the key-value pair and returns nothing.
```

##### vii. Create a second dictionary friend_dict with the same 5 original keys but a friend's (fictional) details. Merge it with your own using the {**dict1, **dict2} syntax, and explain in one line which values “win” when both dictionaries share a key.
